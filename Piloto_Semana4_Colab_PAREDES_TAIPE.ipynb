```python
import os, sys, json, time, glob, hashlib, random, shutil, zipfile, platform, warnings, datetime, io
import numpy as np, pandas as pd
import matplotlib
import matplotlib.pyplot as plt
warnings.filterwarnings("ignore")

SEED = 42                                   # semilla maestra (Protocolo 2.4)
random.seed(SEED); np.random.seed(SEED)
BASE = globals().get("BASE", "/content/piloto_S4")
EVI, FIG, LOG, MOD = f"{BASE}/evidencias", f"{BASE}/figuras", f"{BASE}/logs", f"{BASE}/modelos"
PRIV = f"{BASE}/PRIVADO_no_compartir"       # mapeo de identidades: NO subir ni compartir
for d in [EVI, FIG, LOG, MOD, PRIV, f"{BASE}/real", f"{BASE}/data_raw", f"{BASE}/preprocesado"]:
    os.makedirs(d, exist_ok=True)

BIT, INC = [], []
def _ahora(): return datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
LOGF = f"{LOG}/ejecucion.log"
def reg_log(msg):
    with open(LOGF, "a", encoding="utf-8") as f: f.write(f"[{_ahora()}] {msg}\n")
def bitacora(id_, accion, entrada, esperado, observado, incidencia_txt, evidencia):
    BIT[:] = [b for b in BIT if b["ID"] != id_]
    BIT.append(dict(ID=id_, Hora=_ahora(), Accion=accion, Entrada=entrada, Esperado=esperado, Observado=observado, Incidencia=incidencia_txt, Evidencia=evidencia))
    pd.DataFrame(BIT).to_csv(f"{EVI}/bitacora_ejecucion.csv", index=False, encoding="utf-8-sig")
    reg_log(f"{id_} | {accion} | ESPERADO: {esperado} | OBSERVADO: {observado} | INCIDENCIA: {incidencia_txt} | EVIDENCIA: {evidencia}")
def incidencia(clave, que, por_que, tipo, afecta_validez, decision, evidencia):
    INC[:] = [i for i in INC if i["Clave"] != clave]
    INC.append(dict(Clave=clave, Fecha=_ahora(), Incidencia=que, PorQue=por_que, Tipo=tipo, AfectaValidez=afecta_validez, Decision=decision, Evidencia=evidencia))
    pd.DataFrame(INC).to_csv(f"{EVI}/incidencias.csv", index=False, encoding="utf-8-sig")
    reg_log(f"INCIDENCIA {clave} [{tipo}; afecta validez: {afecta_validez}] {que} -> {decision}")

def versiones():
    v = dict(python=platform.python_version(), numpy=np.__version__, pandas=pd.__version__, fecha=_ahora(), seed_maestra=SEED)
    for mod in ["cv2", "sklearn", "scipy", "PIL", "matplotlib", "tensorflow", "keras"]:
        try: v[mod] = __import__(mod).__version__
        except Exception: v[mod] = "no instalado"
    import subprocess
    try: v["gpu"] = subprocess.run(["nvidia-smi", "--query-gpu=name,memory.total", "--format=csv,noheader"], capture_output=True, text=True, timeout=10).stdout.strip() or "sin GPU"
    except Exception: v["gpu"] = "sin GPU"
    try:
        import psutil; v["ram_total_gb"] = round(psutil.virtual_memory().total / 1e9, 1)
    except Exception: v["ram_total_gb"] = "n/d"
    return v
VERS = versiones()
reg_log("Inicio del piloto Semana 4 | " + json.dumps(VERS, ensure_ascii=False))
json.dump(VERS, open(f"{EVI}/entorno_reproducibilidad.json", "w"), indent=2, ensure_ascii=False)
print(json.dumps(VERS, indent=2, ensure_ascii=False))
```

**Un supuesto que debes confirmar en el validador real:** “ojo izquierdo” = el ojo que aparece a la **izquierda de la imagen** (se confirma con la muestra del Paso 3b).

```python
# ---- Parámetros oficiales del validador SUNEDU + verificador local (pre-filtro, NO decisorio) ----
# ALCANCE V1.4: la CONFORMIDAD se define SOLO con criterios visuales (fondo, ojos/boca, gorra, lentes).
# Peso, formato, dimensiones, dpi y nombre de archivo se registran (c_tecnico_ok) solo como elegibilidad técnica; NO son criterios de conformidad.
import os, io, glob, json, hashlib, time, random, warnings
import numpy as np, pandas as pd, cv2
from PIL import Image, ImageDraw

OFICIAL = dict(
    tam_kb=(4, 50), formato="jpg", ancho=240, alto=288, dpi=300, fondo_rgb=(220, 255),
    ojo_izq=dict(x=(24, 120), y=(55, 180)),
    ojo_der=dict(x=(80, 185), y=(50, 180)),
    boca=dict(x=(50, 161), y=(70, 252)),
)
FOND_UMBRAL = 0.85     # fracción mínima de píxeles de esquinas con R,G,B en [220,255] (calibrado, ver Paso 2)
LENTES_UMBRAL = 0.30   # razón brillo(ojos)/brillo(mejillas) por debajo de la cual se sospecha lente oscuro (calibrado)

_FC = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_alt2.xml")
_EC = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye.xml")
_EC2 = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye_tree_eyeglasses.xml")

def en_rango(v, lo_hi): return lo_hi[0] <= v <= lo_hi[1]

def leer_meta(path):
    """Metadatos técnicos de un archivo (sin decodificar más de lo necesario)."""
    m = dict(tam_kb=os.path.getsize(path) / 1024.0, ext=os.path.splitext(path)[1].lower().lstrip("."))
    with open(path, "rb") as f: head = f.read(8)
    m["es_jpeg_real"] = head[:3] == b"\xff\xd8\xff"
    try:
        with Image.open(path) as im:
            m["ancho"], m["alto"] = im.size
            dpi = im.info.get("dpi", (None, None))
            m["dpi"] = float(dpi[0]) if dpi and dpi[0] else np.nan
            m["modo"] = im.mode
    except Exception:
        m.update(ancho=np.nan, alto=np.nan, dpi=np.nan, modo=None)
    return m

def detectar_rostro(rgb):
    """Rostro + centros de ojos (en coordenadas de la imagen) con Haar de OpenCV. Boca: estimada geométricamente."""
    g = cv2.cvtColor(rgb, cv2.COLOR_RGB2GRAY)
    caras = _FC.detectMultiScale(g, 1.05, 5, minSize=(70, 70))
    if len(caras) == 0: return None
    x, y, w, h = max(caras, key=lambda r: r[2] * r[3])
    roi = g[y:y + int(0.6 * h), x:x + w]
    ojos = _EC.detectMultiScale(roi, 1.05, 4, minSize=(max(8, w // 9), max(8, w // 9)))
    if len(ojos) < 2: ojos = _EC2.detectMultiScale(roi, 1.05, 3, minSize=(max(8, w // 9), max(8, w // 9)))
    izq = [o for o in ojos if x + o[0] + o[2] / 2 < x + w / 2]
    der = [o for o in ojos if x + o[0] + o[2] / 2 >= x + w / 2]
    if not izq or not der: return dict(cara=(x, y, w, h), ojo_izq=None, ojo_der=None, boca=(x + w / 2, y + 0.80 * h))
    oi = max(izq, key=lambda o: o[2] * o[3]); od = max(der, key=lambda o: o[2] * o[3])
    c = lambda o: (x + o[0] + o[2] / 2, y + o[1] + o[3] / 2)
    return dict(cara=(x, y, w, h), ojo_izq=c(oi), ojo_der=c(od), caja_izq=oi, caja_der=od, boca=(x + w / 2, y + 0.80 * h))

def fondo_ok(rgb):
    h, w, _ = rgb.shape
    esq = np.concatenate([rgb[:int(.15 * h), :int(.2 * w)].reshape(-1, 3), rgb[:int(.15 * h), -int(.2 * w):].reshape(-1, 3)])
    lo, hi = OFICIAL["fondo_rgb"]
    frac = float(((esq >= lo) & (esq <= hi)).all(1).mean())
    return frac, frac >= FOND_UMBRAL

def lentes_ratio(rgb, det):
    if det is None or det.get("ojo_izq") is None: return np.nan
    g = cv2.cvtColor(rgb, cv2.COLOR_RGB2GRAY).astype(float)
    (xi, yi), (xd, yd) = det["ojo_izq"], det["ojo_der"]
    r = max(6, int(0.35 * abs(xd - xi) / 2))
    def reg(cx, cy, dy=0): return g[max(0, int(cy + dy - r)):int(cy + dy + r), max(0, int(cx - 1.4 * r)):int(cx + 1.4 * r)]
    ojos = np.mean([reg(xi, yi).mean(), reg(xd, yd).mean()])
    mej = np.mean([reg(xi, yi, 3.2 * r).mean(), reg(xd, yd, 3.2 * r).mean()])
    return float(ojos / max(mej, 1))

def verificar(path):
    """Aplica TODOS los criterios deterministas del validador oficial a un archivo. Devuelve dict con cada criterio (1=cumple)."""
    m = leer_meta(path); r = dict(m)
    r["c_tamano"] = int(en_rango(m["tam_kb"], OFICIAL["tam_kb"]))
    r["c_formato"] = int(m["ext"] in ("jpg", "jpeg") and m["es_jpeg_real"])
    r["c_dimension"] = int(m["ancho"] == OFICIAL["ancho"] and m["alto"] == OFICIAL["alto"])
    r["c_dpi"] = int(m["dpi"] == OFICIAL["dpi"])
    try:
        rgb = np.array(Image.open(path).convert("RGB"))
    except Exception:
        r.update(c_fondo=0, c_ojo_izq=0, c_ojo_der=0, c_boca=0, ratio_lentes=np.nan, frac_fondo=np.nan, rostro_detectado=0); return r
    r["frac_fondo"], ok = fondo_ok(rgb); r["c_fondo"] = int(ok)
    det = detectar_rostro(rgb); r["rostro_detectado"] = int(det is not None)
    if det and det.get("ojo_izq") is not None:
        (xi, yi), (xd, yd), (xb, yb) = det["ojo_izq"], det["ojo_der"], det["boca"]
        r.update(oi_x=xi, oi_y=yi, od_x=xd, od_y=yd, boca_x=xb, boca_y=yb)
        r["c_ojo_izq"] = int(en_rango(xi, OFICIAL["ojo_izq"]["x"]) and en_rango(yi, OFICIAL["ojo_izq"]["y"]))
        r["c_ojo_der"] = int(en_rango(xd, OFICIAL["ojo_der"]["x"]) and en_rango(yd, OFICIAL["ojo_der"]["y"]))
        r["c_boca"] = int(en_rango(xb, OFICIAL["boca"]["x"]) and en_rango(yb, OFICIAL["boca"]["y"]))
    else:
        r.update(c_ojo_izq=0, c_ojo_der=0, c_boca=0)
    r["ratio_lentes"] = lentes_ratio(rgb, det)
    sin_ojos = bool(r["rostro_detectado"]) and np.isnan(r["ratio_lentes"])          # hay rostro pero no se hallan ojos -> sospecha de lente oscuro
    r["c_lentes_heur"] = int(not (sin_ojos or r["ratio_lentes"] < LENTES_UMBRAL))
    r["c_tecnico_ok"] = int(all(r[k] for k in ["c_tamano", "c_formato", "c_dimension", "c_dpi"]))   # informativo: NO entra en la conformidad
    r["cumple_reglas"] = int(all(r[k] for k in ["c_fondo", "c_ojo_izq", "c_ojo_der", "c_boca"]))   # conformidad = solo criterios visuales deterministas
    return r


# ---------- Calibración sobre las fotografías REALES ----------
cal = pd.DataFrame([dict(group_id=g, fuente=f, **verificar(r)) for g, r, f in zip(reales.group_id, reales.ruta_real, reales.fuente)])
crit = ["c_fondo", "c_ojo_izq", "c_ojo_der", "c_boca", "c_lentes_heur", "cumple_reglas", "c_tecnico_ok"]
resumen_cal = cal[crit].mean().mul(100).round(1).rename("% de fotos reales que cumplen")
print(resumen_cal.to_string())
print("\n(c_tecnico_ok = peso/formato/dimensión/dpi: solo informativo, NO es criterio de conformidad del estudio)")
print("Rostro detectado en %d/%d fotos" % (cal.rostro_detectado.sum(), len(cal)))
cal.to_csv(f"{EVI}/calibracion_fotos_reales.csv", index=False)

fig, ax = plt.subplots(1, 3, figsize=(15, 4.2))
for a, (ttl, kx, ky, rng_) in zip(ax, [("Ojo izquierdo (en imagen)", "oi_x", "oi_y", OFICIAL["ojo_izq"]), ("Ojo derecho (en imagen)", "od_x", "od_y", OFICIAL["ojo_der"]), ("Boca (estimada)", "boca_x", "boca_y", OFICIAL["boca"])]):
    a.scatter(cal[kx], cal[ky], s=14, alpha=.7); (x0, x1), (y0, y1) = rng_["x"], rng_["y"]
    a.add_patch(plt.Rectangle((x0, y0), x1 - x0, y1 - y0, fill=False, ec="crimson", lw=2)); a.set_xlim(0, 240); a.set_ylim(288, 0); a.set_title(ttl); a.set_xlabel("X (px)"); a.set_ylabel("Y (px)")
plt.suptitle(f"Posición real de ojos y boca en tus {len(cal)} fotos vs. rango oficial (rectángulo rojo)"); plt.tight_layout(); plt.savefig(f"{FIG}/calibracion_posiciones.png", dpi=130); plt.show()

todas_ok = int(cal.cumple_reglas.sum())
bitacora("P-02", f"Parámetros visuales oficiales en código + calibración del verificador local con las {len(cal)} fotos reales", f"{len(cal)} fotos reales", "Las fotos reales deben cumplir los criterios visuales deterministas (son la referencia de 'Conforme')",
         f"{todas_ok}/{len(cal)} cumplen todas las reglas visuales deterministas; rostro detectado {int(cal.rostro_detectado.sum())}/{len(cal)}; por fuente: {cal.groupby('fuente').cumple_reglas.mean().round(2).to_dict()}", "Supuesto a confirmar en el validador: 'ojo izquierdo' = izquierda de la imagen", "calibracion_fotos_reales.csv, calibracion_posiciones.png")
```

## PASO 3 — Construir el dataset piloto a partir de tus fotos reales (Protocolo P01 · T01–T02)
**Pregunta que hiciste: ¿se pueden modificar las fotografías para que el piloto funcione? → Sí, y es metodológicamente correcto**, siempre que se declare. Las 72 fotos de `Ft.zip` son todas *conformes* (las 15 corregidas pueden traer algún no conforme natural), así que para tener la clase **No conforme** hay que fabricarla. El propio protocolo ya lo prevé (P04: “versiones degradadas… para ampliar la casuística de la clase No conforme”).

**Alcance (decisión tuya, Protocolo V1.4):** la conformidad se define **solo con criterios visuales**. El peso, formato, dimensiones, dpi y el nombre del archivo **no** son criterios; por eso todos los defectos se inyectan **en los píxeles** y cada archivo sigue siendo un JPG de 240×288 a 300 dpi.

**Diseño (reproducible, semilla 42):**
* **Cada estudiante aporta 2 imágenes** (72 de Ft.zip + hasta 15 corregidas): por cada estudiante, 1 imagen *Conforme* (su foto real, re-guardada con el mismo codificador que los defectos para no crear diferencias artificiales) y 1 imagen *No conforme* (su misma foto con **un solo** defecto).
* **4 subtipos de no conformidad** (≈ 21 estudiantes cada uno), alineados con el validador oficial: `fondo` (gris, azul o sombra), `posicion_ojos_boca` (rostro desplazado fuera de los rangos oficiales), `gorra` y `lentes_oscuros` (superpuestos digitalmente).
* El *group_id* se comparte entre las 2 imágenes del mismo estudiante → esto es lo que hará **necesaria la partición agrupada** del Paso 5.
* Se añaden **6 “trampas” de auditoría** (3 duplicados exactos y 3 archivos corruptos de 3 maneras distintas) para comprobar que el Paso 4 las detecta. La lista de trampas se guarda aparte.

> **Qué NO es:** este *ground truth* se conoce por construcción; **no** viene del validador oficial hasta que hagas el Paso 3b–3c. Los defectos son sintéticos (p. ej. la gorra es un dibujo): este piloto prueba la **mecánica** del flujo, no el desempeño científico final.


```python
# ---- Generador de defectos controlados sobre fotografías REALES (semilla fija) ----
import numpy as np, cv2, io, os
from PIL import Image, ImageDraw, ImageFilter

SUBTIPOS = ["fondo", "posicion_ojos_boca", "gorra", "lentes_oscuros"]   # alcance V1.4: SOLO criterios visuales
FAMILIA = dict(fondo="visual", posicion_ojos_boca="visual", gorra="visual", lentes_oscuros="visual")
CRITERIO = dict(fondo="c_fondo", posicion_ojos_boca=("c_ojo_izq", "c_ojo_der", "c_boca"), gorra=None, lentes_oscuros=None)

def guardar_jpeg_conforme(img, path, dpi=300, kb_max=49.0):
    """Guarda JPG con dpi indicado y la MAYOR calidad que mantenga el peso <= kb_max (para no violar el criterio de 4-50 KB sin querer)."""
    for q in range(100, 20, -2):
        buf = io.BytesIO(); img.save(buf, "JPEG", quality=q, subsampling=0, dpi=(dpi, dpi))
        if buf.tell() / 1024 <= kb_max: break
    with open(path, "wb") as f: f.write(buf.getvalue())
    return q

def mascara_fondo(rgb):
    """Fondo = píxeles claros conectados con el borde superior/lateral (heurística simple, suficiente para el piloto)."""
    h, w, _ = rgb.shape
    cand = (rgb.min(2) >= 190).astype(np.uint8)
    n, lab = cv2.connectedComponents(cand, connectivity=8)
    keep = set(np.unique(lab[0, :])) | set(np.unique(lab[:int(.6 * h), 0])) | set(np.unique(lab[:int(.6 * h), -1]))
    keep.discard(0)
    m = np.isin(lab, list(keep)).astype(np.float32)
    return cv2.GaussianBlur(m, (5, 5), 0)

def d_fondo(rgb, rng, det, var):
    m = mascara_fondo(rgb)[..., None]; h, w, _ = rgb.shape
    if var == 0:   nuevo = np.full_like(rgb, (165, 165, 165), dtype=np.uint8)                    # gris
    elif var == 1: nuevo = np.full_like(rgb, (140, 175, 225), dtype=np.uint8)                    # azul
    else:                                                                                         # degradado/sombra
        g = np.linspace(255, 110, w)[None, :, None].repeat(h, 0); nuevo = np.concatenate([g, g, g], 2).astype(np.uint8)
    return np.clip(m * nuevo + (1 - m) * rgb, 0, 255).astype(np.uint8)

def d_posicion(rgb, rng, det, var):
    dx, dy = [(55, 0), (-80, 0), (0, 80), (0, -75)][var % 4]
    M = np.float32([[1, 0, dx], [0, 1, dy]])
    return cv2.warpAffine(rgb, M, (rgb.shape[1], rgb.shape[0]), borderMode=cv2.BORDER_CONSTANT, borderValue=(255, 255, 255))

def _overlay(rgb, draw_fn, ss=4):
    h, w, _ = rgb.shape
    cap = Image.new("RGBA", (w * ss, h * ss), (0, 0, 0, 0)); d = ImageDraw.Draw(cap); draw_fn(d, ss)
    cap = cap.resize((w, h), Image.LANCZOS)
    base = Image.fromarray(rgb).convert("RGBA"); base.alpha_composite(cap)
    return np.array(base.convert("RGB"))

def d_gorra(rgb, rng, det, var):
    x, y, fw, fh = det["cara"]; col = [(25, 25, 35), (30, 50, 110), (150, 25, 25), (60, 60, 60)][var % 4]
    def f(d, s):
        cx = (x + fw / 2) * s; top = (y - 0.38 * fh) * s; bot = (y + 0.20 * fh) * s; rw = 0.62 * fw * s
        d.pieslice([cx - rw, top, cx + rw, bot + (bot - top)], 180, 360, fill=col + (255,))
        d.ellipse([cx - 0.80 * fw * s, bot - 0.05 * fh * s, cx + 0.80 * fw * s, bot + 0.07 * fh * s], fill=tuple(int(c * .7) for c in col) + (255,))
    return _overlay(rgb, f)

def d_lentes(rgb, rng, det, var):
    (xi, yi), (xd, yd) = det["ojo_izq"], det["ojo_der"]; _, _, fw, fh = det["cara"]
    ww, hh = 0.30 * fw, 0.17 * fh; col = [(15, 15, 20), (30, 25, 20)][var % 2]
    def f(d, s):
        for cx, cy in [(xi, yi), (xd, yd)]:
            d.rounded_rectangle([(cx - ww / 2) * s, (cy - hh / 2) * s, (cx + ww / 2) * s, (cy + hh / 2) * s], radius=int(0.05 * fw * s), fill=col + (250,))
        d.line([((xi + ww / 2) * s, yi * s), ((xd - ww / 2) * s, yd * s)], fill=col + (255,), width=int(0.025 * fw * s))
        d.line([((xi - ww / 2) * s, yi * s), ((xi - ww / 2 - 0.15 * fw) * s, (yi - 0.02 * fh) * s)], fill=col + (255,), width=int(0.02 * fw * s))
        d.line([((xd + ww / 2) * s, yd * s), ((xd + ww / 2 + 0.15 * fw) * s, (yd - 0.02 * fh) * s)], fill=col + (255,), width=int(0.02 * fw * s))
    return _overlay(rgb, f)

def generar_defecto(src_path, subtipo, var, out_stem, detectar_rostro):
    """Devuelve (ruta_salida, info). Cada defecto viola UN solo criterio (el resto permanece dentro de la norma)."""
    img = Image.open(src_path).convert("RGB"); rgb = np.array(img); info = dict(subtipo=subtipo, variante=var)
    if subtipo in ("fondo", "posicion_ojos_boca", "gorra", "lentes_oscuros"):
        det = detectar_rostro(rgb)
        fn = dict(fondo=d_fondo, posicion_ojos_boca=d_posicion, gorra=d_gorra, lentes_oscuros=d_lentes)[subtipo]
        out = Image.fromarray(fn(rgb, None, det, var)); p = out_stem + ".jpg"; guardar_jpeg_conforme(out, p); return p, info
    raise ValueError(subtipo)


# ---------- Construcción del dataset piloto ----------
shutil.rmtree(f"{BASE}/data_raw", ignore_errors=True); os.makedirs(f"{BASE}/data_raw")
ULT_ID = f"IMG_{2 * len(reales):04d}"                              # última imagen del dataset; las siguientes son trampas de auditoría
fuente_g = dict(zip(reales.group_id, reales.fuente))
rng = np.random.RandomState(SEED)
asign = np.array(SUBTIPOS * (len(reales) // len(SUBTIPOS) + 1))[:len(reales)]; rng.shuffle(asign)
orden_ids = rng.permutation(2 * len(reales)) + 1
ctr = {s: 0 for s in SUBTIPOS}; filas = []; verif = {}
for i, (g, rpath) in enumerate(zip(reales.group_id, reales.ruta_real)):
    # --- imagen CONFORME (foto real re-guardada con el mismo codificador) ---
    iid = f"IMG_{orden_ids[2*i]:04d}"; p = f"{BASE}/data_raw/{iid}.jpg"
    guardar_jpeg_conforme(Image.open(rpath).convert("RGB"), p)
    v = verificar(p); verif[iid] = v
    if v["cumple_reglas"] and v["c_lentes_heur"]:
        clase, sub, fam, org = 1, "ninguno", "-", "real_recodificada"
    else:                                   # hallazgo natural: una foto real que NO pasa las reglas -> se reclasifica y se documenta
        falla = [k for k in ["c_fondo", "c_ojo_izq", "c_ojo_der", "c_boca", "c_lentes_heur"] if not v[k]]
        clase, sub, fam, org = 0, "natural_" + falla[0], "visual", "real_no_conforme_natural"
    filas.append(dict(image_id=iid, file_name=os.path.basename(p), group_id=g, clase=clase, subtipo=sub, familia=fam, origen=org))
    if clase == 0: continue                  # ya es no conforme por sí misma: no se le añade un 2.º defecto (un solo defecto por imagen)
    # --- imagen NO CONFORME (misma foto + 1 defecto) ---
    s = asign[i]; var = ctr[s] % (3 if s == "fondo" else 4 if s == "posicion_ojos_boca" else 2); ctr[s] += 1
    iid2 = f"IMG_{orden_ids[2*i+1]:04d}"
    p2, info = generar_defecto(rpath, s, var, f"{BASE}/data_raw/{iid2}", detectar_rostro)
    verif[iid2] = verificar(p2)
    filas.append(dict(image_id=iid2, file_name=os.path.basename(p2), group_id=g, clase=0, subtipo=s, familia=FAMILIA[s], origen="real_con_defecto_sintetico", detalle_defecto=info.get("detalle", "")))
meta = pd.DataFrame(filas).sort_values("image_id").reset_index(drop=True)
meta["etiqueta_fuente"] = "piloto_por_construccion (NO validador oficial)"
meta["fuente"] = meta.group_id.map(fuente_g)
meta["source"] = meta.fuente
for k, c in [("dpi", "dpi"), ("ancho", "width"), ("alto", "height"), ("tam_kb", "tam_kb"), ("ext", "formato")]:
    meta[c] = [verif[i][k] for i in meta.image_id]
meta["tam_kb"] = meta.tam_kb.round(2)

# ---------- Trampas de auditoría (se guardan APARTE) ----------
trampas = []; nxt = 2 * len(reales) + 1
conf = meta[meta.clase == 1].sample(3, random_state=SEED); jpgs = meta[(meta.clase == 0) & meta.file_name.str.endswith(".jpg")].sample(3, random_state=SEED)
for _, r in conf.iterrows():                                              # 3 duplicados exactos
    iid = f"IMG_{nxt:04d}"; nxt += 1; shutil.copyfile(f"{BASE}/data_raw/{r.file_name}", f"{BASE}/data_raw/{iid}.jpg")
    trampas.append((iid, "duplicado_exacto", r.image_id)); meta = pd.concat([meta, pd.DataFrame([{**r.to_dict(), "image_id": iid, "file_name": iid + ".jpg"}])], ignore_index=True)
for (_, r), modo in zip(jpgs.iterrows(), ["truncado", "bytes_corruptos_medio", "cabecera_destruida"]):   # 3 corruptos distintos
    iid = f"IMG_{nxt:04d}"; nxt += 1; b = bytearray(open(f"{BASE}/data_raw/{r.file_name}", "rb").read())
    if modo == "truncado": b = b[:int(len(b) * .6)]
    elif modo == "bytes_corruptos_medio":
        for k in range(len(b) // 2, len(b) // 2 + 400): b[k] = random.randrange(256)
    else:
        for k in range(0, 300): b[k] = random.randrange(256)
    open(f"{BASE}/data_raw/{iid}.jpg", "wb").write(bytes(b)); trampas.append((iid, "corrupto_" + modo, r.image_id))
    meta = pd.concat([meta, pd.DataFrame([{**r.to_dict(), "image_id": iid, "file_name": iid + ".jpg"}])], ignore_index=True)
pd.DataFrame(trampas, columns=["image_id", "trampa", "copiado_de"]).to_csv(f"{PRIV}/trampas_auditoria_VERDAD.csv", index=False)
meta["status"], meta["reason"] = "registrado", ""
meta.to_csv(f"{EVI}/dataset_metadata.csv", index=False, encoding="utf-8-sig")
print("Archivos en data_raw:", len(meta), f"({int((meta.image_id <= ULT_ID).sum())} del dataset + 6 trampas)")
print(meta[meta.image_id <= ULT_ID].groupby(["fuente", "clase"]).size().to_string())

# ---------- ¿Las reglas deterministas ven cada defecto? (concordancia verificador vs. construcción) ----------
V = pd.DataFrame(verif).T.reset_index(names="image_id").merge(meta[["image_id", "clase", "subtipo", "familia"]].drop_duplicates("image_id"), on="image_id")
def detecta(r):
    s = r.subtipo
    if s == "ninguno": return int(r.cumple_reglas == 0 or r.c_lentes_heur == 0)          # falsa alarma
    if s.startswith("natural_"): return int(r.cumple_reglas == 0 or r.c_lentes_heur == 0)   # no conforme natural: la regla local lo marcó (así se etiquetó)
    if s == "posicion_ojos_boca": return int(not (r.c_ojo_izq and r.c_ojo_der and r.c_boca))
    if s == "lentes_oscuros": return int(r.c_lentes_heur == 0)
    if s == "gorra": return np.nan                                                        # sin regla determinista posible
    return int(r[CRITERIO[s]] == 0)
V["detectado_por_regla"] = V.apply(detecta, axis=1)
conc = V.groupby("subtipo").detectado_por_regla.agg(["size", "mean"]).rename(columns={"size": "n", "mean": "detecta_regla"}).round(2)
conc.loc["gorra", "detecta_regla"] = np.nan
print("\nFracción de imágenes que el verificador local marca por la regla correspondiente ('ninguno' = falsas alarmas; 'gorra' = sin regla posible):")
print(conc.to_string()); conc.to_csv(f"{EVI}/concordancia_verificador_vs_construccion.csv")

ej = {s: meta[(meta.subtipo == s) & (meta.image_id <= ULT_ID)].iloc[0] for s in SUBTIPOS}
fig, ax = plt.subplots(2, 5, figsize=(14, 7.2)); ax = ax.ravel()
c0 = meta[(meta.clase == 1) & (meta.image_id <= ULT_ID)].iloc[0]; ax[0].imshow(Image.open(f"{BASE}/data_raw/{c0.file_name}")); ax[0].set_title("CONFORME (real)")
for a, s in zip(ax[1:], SUBTIPOS): a.imshow(Image.open(f"{BASE}/data_raw/{ej[s].file_name}").convert("RGB")); a.set_title(s)
for a in ax: a.axis("off")
ax[9].axis("off"); plt.tight_layout(); plt.savefig(f"{FIG}/ejemplos_por_subtipo.png", dpi=120); plt.show()
bitacora("P-03", f"Construir dataset piloto ({int((meta.image_id <= ULT_ID).sum())} imágenes) desde las {len(reales)} fotos reales + 6 trampas de auditoría", f"{len(reales)} fotos reales", "1 conforme + 1 con defecto visual único por estudiante (los no conformes naturales no reciben 2.º defecto) + 6 trampas",
         f"{int((meta.image_id<=ULT_ID).sum())} imágenes del dataset (clase1={int(((meta.clase==1)&(meta.image_id<=ULT_ID)).sum())}) + {len(trampas)} trampas; {int((meta.origen=='real_no_conforme_natural').sum())} fotos reales no pasaron las reglas locales (no conformes naturales)",
         "Ninguna" , "dataset_metadata.csv, concordancia_verificador_vs_construccion.csv, ejemplos_por_subtipo.png")
```

```python
# ---------- PASO 3b: muestra de calibración para el validador oficial ----------
meta_ = pd.read_csv(f"{EVI}/dataset_metadata.csv")
base_ = meta_[meta_.image_id <= ULT_ID]
rs_ = np.random.RandomState(SEED)
nat_nc = list(base_[base_.origen == "real_no_conforme_natural"].image_id)[:3]                       # no conformes REALES (los más valiosos)
cand_c = base_[(base_.origen == "real_recodificada") & (base_.fuente != "Ft.zip")]
nat_c = list(cand_c.sample(min(2, len(cand_c)), random_state=SEED).image_id)                         # conformes reales de la fuente adicional
cand_f = base_[(base_.origen == "real_recodificada") & (base_.fuente == "Ft.zip")]
cal_sel = nat_nc + nat_c + list(cand_f.sample(2, random_state=SEED).image_id)
for s_, k_ in [("fondo", 1), ("posicion_ojos_boca", 1), ("gorra", 3), ("lentes_oscuros", 3)]:       # gorra y lentes: la duda clave (¿el validador los rechaza?)
    cand = base_[base_.subtipo == s_]; cal_sel += list(cand.sample(min(k_, len(cand)), random_state=SEED).image_id)
cal_sel = cal_sel[:15]                                                                                # el protocolo pide 10–15
CAL = f"{BASE}/calibracion_oficial"; shutil.rmtree(CAL, ignore_errors=True); os.makedirs(CAL)
filas = []
for k, iid in enumerate(rs_.permutation(cal_sel), start=1):
    r = base_[base_.image_id == iid].iloc[0]; nombre = f"CAL_{k:02d}" + os.path.splitext(r.file_name)[1]
    shutil.copyfile(f"{BASE}/data_raw/{r.file_name}", f"{CAL}/{nombre}")
    filas.append(dict(archivo=nombre, resultado_oficial=("No aplica" if nombre.endswith(".png") else ""), motivo_oficial="", fecha_validacion="", image_id_PRIVADO=iid, subtipo_construido=r.subtipo, origen_PRIVADO=r.origen))
plantilla = pd.DataFrame(filas).sort_values("archivo")
plantilla.drop(columns=["image_id_PRIVADO", "subtipo_construido", "origen_PRIVADO"]).to_csv(f"{CAL}/sunedu_validation_log_PLANTILLA.csv", index=False, encoding="utf-8-sig")
plantilla.to_csv(f"{PRIV}/calibracion_clave_PRIVADA.csv", index=False)         # clave: NO se la muestres al validador ni a ti mismo antes de validar
zc = f"{BASE}/CAL_para_validador.zip"
with zipfile.ZipFile(zc, "w") as z:
    for f in sorted(glob.glob(f"{CAL}/*")): z.write(f, os.path.basename(f))
print(f"Muestra de calibración: {len(plantilla)} archivos ({int((plantilla.archivo.str.endswith('.jpg')).sum())} subibles al validador).")
try:
    from google.colab import files; files.download(zc)
except Exception: print("ZIP en:", zc)
```

## PASO 3c — Leer el resultado oficial y decidir qué subtipos son realmente "No conforme"
**Qué hace:** lee tu `sunedu_validation_log.csv` y compara, **imagen por imagen**, lo que dijo el validador oficial con lo que construimos. Después decide por subtipo:
* Si el validador **rechaza** el subtipo → se mantiene como No conforme.
* Si el validador lo **acepta** (p. ej. una gorra dibujada) → ese subtipo **no puede** ser No conforme en el piloto: sus imágenes se **excluyen** del dataset con motivo `aceptada_por_validador_oficial` (la etiqueta manda el validador, no nosotros) y se registra una incidencia.
* Si el validador oficial **contradice la etiqueta local de una foto real** (p. ej. una foto con lentes que el verificador marcó como no conforme), esa foto se **reetiqueta** según el oficial y se registra una incidencia.
* Muestra la **concordancia del verificador local** con el validador (la calibración que pide el Protocolo V1.2).

Si todavía no tienes el CSV, la celda **continúa** sin exclusiones y lo deja anotado: pero entonces el piloto queda con etiqueta "por construcción" y el semáforo no podrá ser VERDE.

## PASO 4 — Auditoría de calidad del dataset (Protocolo P02 · T03)
**Qué hace:** revisa los 150 archivos como lo haría el auditor del estudio real:
1. **Archivos corruptos** — se compara `PIL verify()` **solo** contra `verify()` **+ `load()`** (decodificación completa).
2. **Duplicados exactos** — hash SHA-256 de los bytes.
3. **Criterios visuales del validador** — fondo (R,G,B 220–255) y posición de ojos y boca (6 rangos), más la heurística de lentes oscuros.
4. **Exclusiones** según la sección 2.2 del Protocolo V1.4: duplicado, corrupto, sin etiqueta y (si el Paso 3c lo indica) subtipo aceptado por el validador oficial. *El peso, formato, dimensiones, dpi y nombre del archivo ya NO se auditan como criterio: están fuera del alcance de la conformidad.*

**Qué esperamos observar:** las 6 trampas detectadas y 0 falsos positivos; si `verify()` solo se queda corto, es un hallazgo que corrige el Protocolo (P01 dice que `verify()` detecta el 100 %).


```python
# ---------- PASO 4: auditoría ----------
def integridad(path):
    """Devuelve (falla_solo_verify, falla_verify_y_load)."""
    f1 = f2 = False
    try:
        with Image.open(path) as im: im.verify()
    except Exception: f1 = True
    try:
        with Image.open(path) as im: im.verify()
        with Image.open(path) as im: im.load()
    except Exception: f2 = True
    return f1, f2
def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f: h.update(f.read())
    return h.hexdigest()

meta = pd.read_csv(f"{EVI}/dataset_metadata.csv").sort_values("image_id").reset_index(drop=True)
for _iid, _c in globals().get("RELABEL", {}).items():                      # el validador oficial manda sobre el verificador local
    _m = meta.image_id == _iid
    meta.loc[_m, "clase"] = _c; meta.loc[_m, "subtipo"] = "ninguno" if _c == 1 else "oficial_no_conforme"; meta.loc[_m, "familia"] = "-" if _c == 1 else "visual"; meta.loc[_m, "origen"] = "real_reetiquetada_oficial"
meta.loc[meta.image_id.isin(globals().get("OFICIAL_IDS", set())), "etiqueta_fuente"] = "validador_oficial"
aud = []; vistos = {}
for _, r in meta.iterrows():
    p = f"{BASE}/data_raw/{r.file_name}"; f1, f2 = integridad(p); h = sha256(p)
    fila = dict(image_id=r.image_id, sha256=h, corrupto_solo_verify=int(f1), corrupto_verify_y_load=int(f2), duplicado_de=vistos.get(h, ""))
    vistos.setdefault(h, r.image_id)
    if not f2: fila.update({k: v for k, v in verificar(p).items() if k.startswith("c_") or k in ("tam_kb", "ancho", "alto", "dpi", "cumple_reglas", "frac_fondo", "rostro_detectado", "ratio_lentes")})
    aud.append(fila)
aud = pd.DataFrame(aud).merge(meta[["image_id", "group_id", "clase", "subtipo", "familia", "width", "height", "dpi"]].rename(columns={"width": "w_reg", "height": "h_reg", "dpi": "dpi_reg"}), on="image_id")

def motivo(r):
    if r.image_id in globals().get("EXCLUIR_OFICIAL", set()): return "aceptada_por_validador_oficial"
    if r.corrupto_verify_y_load: return "corrupto"
    if r.duplicado_de: return "duplicado_exacto"
    if pd.isna(r.clase): return "sin_etiqueta"
    # V1.4: los metadatos de archivo (dpi, tamaño, formato, dimensiones) NO son criterio de inclusión/exclusión ni de conformidad
    return ""
aud["motivo_exclusion"] = aud.apply(motivo, axis=1); aud["incluido"] = (aud.motivo_exclusion == "").astype(int)
aud.to_csv(f"{EVI}/dataset_audit.csv", index=False)

verdad = pd.read_csv(f"{PRIV}/trampas_auditoria_VERDAD.csv")
t = verdad.merge(aud[["image_id", "corrupto_solo_verify", "corrupto_verify_y_load", "duplicado_de", "motivo_exclusion"]], on="image_id")
corr = t[t.trampa.str.startswith("corrupto")]; dup = t[t.trampa == "duplicado_exacto"]
res = dict(archivos_auditados=len(aud), corruptos_insertados=len(corr), detectados_solo_verify=int(corr.corrupto_solo_verify.sum()), detectados_verify_y_load=int(corr.corrupto_verify_y_load.sum()),
           duplicados_insertados=len(dup), duplicados_detectados=int((dup.duplicado_de != "").sum()),
           falsos_positivos_exclusion=int(((aud.motivo_exclusion != "") & (aud.motivo_exclusion != "aceptada_por_validador_oficial") & ~aud.image_id.isin(verdad.image_id)).sum()),
           incluidos=int(aud.incluido.sum()), excluidos=int((aud.incluido == 0).sum()), exclusiones_por_motivo=aud[aud.incluido == 0].motivo_exclusion.value_counts().to_dict())
json.dump(res, open(f"{EVI}/resumen_auditoria.json", "w"), indent=2, ensure_ascii=False)
print(json.dumps(res, indent=2, ensure_ascii=False)); print(t[["image_id", "trampa", "corrupto_solo_verify", "corrupto_verify_y_load", "motivo_exclusion"]].to_string(index=False))

clean = meta.merge(aud[["image_id", "incluido", "motivo_exclusion", "cumple_reglas", "c_tamano", "c_formato", "c_dimension", "c_dpi", "c_fondo", "c_ojo_izq", "c_ojo_der", "c_boca", "c_lentes_heur", "ratio_lentes"]], on="image_id")
meta["status"] = np.where(clean.incluido == 1, "incluido", "excluido"); meta["reason"] = clean.motivo_exclusion.values
meta.to_csv(f"{EVI}/dataset_metadata.csv", index=False, encoding="utf-8-sig")
clean = clean[clean.incluido == 1].reset_index(drop=True); clean.to_csv(f"{EVI}/dataset_clean.csv", index=False, encoding="utf-8-sig")
print("\nDataset limpio:", len(clean), "imágenes |", clean.group_id.nunique(), "estudiantes | % conformes =", round(100 * clean.clase.mean(), 1))
print(clean.groupby("subtipo").size().to_string())

res["excluidas_por_validador_oficial"] = int((aud.motivo_exclusion == "aceptada_por_validador_oficial").sum()); json.dump(res, open(f"{EVI}/resumen_auditoria.json", "w"), indent=2, ensure_ascii=False)
ok = (res["detectados_verify_y_load"] == res["corruptos_insertados"]) and (res["duplicados_detectados"] == res["duplicados_insertados"]) and res["falsos_positivos_exclusion"] == 0
bitacora("P-04", "Auditar 150 archivos: corruptos (verify vs verify+load), duplicados SHA-256, exclusiones", "150 archivos de data_raw/", "Detectar 3/3 corruptos, 3/3 duplicados, 0 falsos positivos",
         f"corruptos: verify solo {res['detectados_solo_verify']}/3, verify+load {res['detectados_verify_y_load']}/3; duplicados {res['duplicados_detectados']}/3; falsos positivos {res['falsos_positivos_exclusion']}; dataset limpio = {len(clean)}",
         "verify() por sí solo NO detecta archivos truncados o corruptos en el interior" if res["detectados_solo_verify"] < 3 else "Ninguna", "dataset_audit.csv, resumen_auditoria.json, dataset_clean.csv")
if res["detectados_solo_verify"] < res["corruptos_insertados"]:
    incidencia("INC-VERIFY", f"PIL verify() detectó solo {res['detectados_solo_verify']} de {res['corruptos_insertados']} archivos corruptos; verify()+load() detectó {res['detectados_verify_y_load']}",
               "verify() solo revisa la estructura/cabecera; no decodifica los píxeles, así que no ve truncamientos ni daños en el interior del archivo", "Técnica", "Sí",
               "Corregir P01 del Protocolo: método = verify() + reabrir + load() (decodificación completa).", "resumen_auditoria.json, dataset_audit.csv")
assert ok, "La auditoría no cumplió lo esperado: revisa resumen_auditoria.json antes de seguir"
```

## PASO 5 — Partición 70/15/15 y control de fuga de datos (Protocolo P03 · T04–T05)
**Qué hace, en dos intentos (como en el protocolo):**
* **Intento 1 — partición aleatoria simple.** Sirve para *medir* el problema: como cada estudiante tiene 2 imágenes (una conforme y otra con defecto), una partición ingenua pone a veces una en *train* y otra en *test*: la red habría “visto” a esa persona antes de evaluarla (**fuga de datos**).
* **Intento 2 — partición agrupada por `group_id`** con `GroupShuffleSplit` en dos etapas (70 % vs 30 %; luego el 30 % en 15 %/15 %), semilla 42.

**Verificación OBLIGATORIA (P03 V1.2):** intersección de `group_id` entre train∩val, train∩test y val∩test = ∅. Si no es vacía, el cuaderno **se detiene**.

> Nota: `GroupShuffleSplit` agrupa pero **no estratifica**. Aquí el balance 50/50 sale perfecto porque cada estudiante aporta 1 conforme + 1 no conforme; con el dataset real (≈1000 fotos) ya no será automático, por eso la celda también **verifica el balance** (tolerancia ±5 puntos).


```python
# ---------- PASO 5: partición ----------
from sklearn.model_selection import GroupShuffleSplit
df = pd.read_csv(f"{EVI}/dataset_clean.csv")

# Intento 1: aleatoria simple (para MEDIR la fuga)
rs = np.random.RandomState(SEED); idx = rs.permutation(len(df)); n_tr, n_va = int(.70 * len(df)), int(.15 * len(df))
ingenua = np.array(["train"] * len(df), dtype=object); ingenua[idx[n_tr:n_tr + n_va]] = "val"; ingenua[idx[n_tr + n_va:]] = "test"
n_part = df.assign(split=ingenua).groupby("group_id").split.nunique()
fuga_ingenua = int((n_part > 1).sum())
print(f"INTENTO 1 (aleatoria simple): {fuga_ingenua} de {df.group_id.nunique()} estudiantes aparecen en MÁS DE UNA partición  → FUGA DE DATOS")
json.dump(dict(metodo="aleatoria simple", semilla=SEED, estudiantes_en_mas_de_una_particion=fuga_ingenua, estudiantes_total=int(df.group_id.nunique())), open(f"{EVI}/incidencia_particion_ingenua.json", "w"), indent=2)

# Intento 2: agrupada, 2 etapas
g1 = GroupShuffleSplit(n_splits=1, test_size=0.30, random_state=SEED); i_tr, i_rest = next(g1.split(df, groups=df.group_id))
rest = df.iloc[i_rest]; g2 = GroupShuffleSplit(n_splits=1, test_size=0.50, random_state=SEED); i_va, i_te = next(g2.split(rest, groups=rest.group_id))
df["split"] = ""; df.loc[df.index[i_tr], "split"] = "train"; df.loc[rest.index[i_va], "split"] = "val"; df.loc[rest.index[i_te], "split"] = "test"
G = {s: set(df[df.split == s].group_id) for s in ["train", "val", "test"]}
sol = dict(train_val=len(G["train"] & G["val"]), train_test=len(G["train"] & G["test"]), val_test=len(G["val"] & G["test"]))
bal = df.groupby("split").clase.agg(n="size", pct_conforme=lambda s: round(100 * s.mean(), 1)); print("\nINTENTO 2 (agrupada): solapamientos =", sol); print(bal.to_string())
ok_bal = bool(((bal.pct_conforme - 50).abs() <= 5).all())
json.dump(dict(algoritmo="GroupShuffleSplit 2 etapas (70/30 y 15/15), semilla 42", solapamientos=sol, solapamiento_total=sum(sol.values()), balance_por_particion=bal.to_dict(), balance_dentro_de_5pp=ok_bal,
               fuga_con_particion_ingenua=fuga_ingenua), open(f"{EVI}/verificacion_solapamiento.json", "w"), indent=2)
assert sum(sol.values()) == 0, "FUGA: el protocolo PROHÍBE continuar. Corrige la partición."
df[["image_id", "group_id", "clase", "subtipo", "familia", "split"]].to_csv(f"{EVI}/dataset_split.csv", index=False)
df.to_csv(f"{EVI}/dataset_clean.csv", index=False, encoding="utf-8-sig")
ax = df.groupby(["split", "clase"]).size().unstack().reindex(["train", "val", "test"]).rename(columns={0: "No conforme", 1: "Conforme"}).plot.bar(figsize=(5.5, 3.6), rot=0); ax.set_title("Imágenes por partición y clase"); plt.tight_layout(); plt.savefig(f"{FIG}/particion.png", dpi=120); plt.show()
bitacora("P-05", "Partición 70/15/15: (1) aleatoria simple y (2) agrupada por group_id (GroupShuffleSplit 2 etapas, seed 42) + verificación de solapamiento", f"{len(df)} imágenes / {df.group_id.nunique()} estudiantes", "Solapamiento = 0; balance 50/50 por partición",
         f"Intento 1: {fuga_ingenua} estudiantes en >1 partición (fuga). Intento 2: solapamientos {sol}; train/val/test = {bal.n.to_dict()}; balance ok={ok_bal}",
         "Fuga de datos con partición aleatoria simple (confirmada)", "incidencia_particion_ingenua.json, verificacion_solapamiento.json, dataset_split.csv")
incidencia("INC-FUGA", f"Partición aleatoria simple: {fuga_ingenua} estudiantes con imágenes en más de una partición", "Cada estudiante aporta varias imágenes; una partición por imagen no respeta al estudiante", "Metodológica", "Sí",
           "Mantener partición agrupada (GroupShuffleSplit) + verificación de solapamiento = 0 como paso obligatorio (ya en V1.2).", "incidencia_particion_ingenua.json, verificacion_solapamiento.json")
```

## PASO 6 — Preprocesamiento (Protocolo P04 · T06)
**Qué hace, en este orden (idéntico al protocolo):** lectura → conversión RGB → **redimensionar a 224×224×3** → **guardar como JPG calidad 90** (parámetro fijado en V1.1) → releer → **normalizar a [0,1]** (÷255). El *data augmentation* **solo** se aplica al entrenamiento: rotación ±10°, brillo ±20 %, contraste ±15 %, volteo horizontal (p = 0.5). Validación y prueba **nunca** reciben augmentation.

**Coherencia con el alcance (V1.4):** al redimensionar a 224×224 y recomprimir se pierden el peso, formato, dimensiones originales y dpi del archivo. Eso **no es un problema** aquí: esos metadatos están fuera de la conformidad del estudio. La red recibe solo los píxeles, que es justo lo que decide la conformidad visual.

```python
# ---------- PASO 6: preprocesamiento + augmentation ----------
PRE = f"{BASE}/preprocesado"; shutil.rmtree(PRE, ignore_errors=True)
df = pd.read_csv(f"{EVI}/dataset_clean.csv")
for s in ["train", "val", "test"]: os.makedirs(f"{PRE}/{s}", exist_ok=True)
def preprocesar(path, out):
    im = Image.open(path).convert("RGB").resize((224, 224), Image.BILINEAR)
    im.save(out, "JPEG", quality=90)                                      # parámetro uniforme V1.1
    return np.asarray(Image.open(out).convert("RGB"), dtype=np.float32) / 255.0
D = {}
for s in ["train", "val", "test"]:
    sub = df[df.split == s].reset_index(drop=True)
    X = np.stack([preprocesar(f"{BASE}/data_raw/{r.file_name}", f"{PRE}/{s}/{r.image_id}.jpg") for _, r in sub.iterrows()])
    D[s] = dict(X=X, y=sub.clase.values.astype("float32"), meta=sub)
    print(f"{s:5s} X={X.shape} rango=[{X.min():.2f},{X.max():.2f}]  %conforme={100*sub.clase.mean():.1f}")

AUG = dict(rot_grados=10, brillo=0.20, contraste=0.15, flip_horizontal_p=0.5)
def augmentar(x, rng):
    """x en [0,1], (224,224,3). Solo para ENTRENAMIENTO."""
    if rng.rand() < AUG["flip_horizontal_p"]: x = x[:, ::-1]
    a = rng.uniform(-AUG["rot_grados"], AUG["rot_grados"]); M = cv2.getRotationMatrix2D((112, 112), a, 1.0)
    x = cv2.warpAffine(np.ascontiguousarray(x), M, (224, 224), borderMode=cv2.BORDER_REPLICATE)
    x = x * rng.uniform(1 - AUG["brillo"], 1 + AUG["brillo"])
    m = x.mean(); x = (x - m) * rng.uniform(1 - AUG["contraste"], 1 + AUG["contraste"]) + m
    return np.clip(x, 0, 1).astype("float32")
json.dump(AUG | dict(jpg_calidad=90, tamano=[224, 224, 3], normalizacion="x/255 -> [0,1]", solo_en="train"), open(f"{EVI}/parametros_preprocesamiento.json", "w"), indent=2)

r0 = df[df.split == "train"].iloc[0]; rr = np.random.RandomState(0)
fig, ax = plt.subplots(1, 6, figsize=(15, 3)); ax[0].imshow(Image.open(f"{BASE}/data_raw/{r0.file_name}").convert("RGB")); ax[0].set_title("Original")
x0 = D["train"]["X"][0]; ax[1].imshow(x0); ax[1].set_title("224×224, JPG q90")
for k in range(4): ax[2 + k].imshow(augmentar(x0, rr)); ax[2 + k].set_title(f"Aug {k+1} (solo train)")
for a in ax: a.axis("off")
plt.tight_layout(); plt.savefig(f"{FIG}/preprocesamiento_antes_despues.png", dpi=120); plt.show()
np.savez_compressed(f"{MOD}/datos_preprocesados.npz", **{f"{s}_{k}": D[s][k] for s in D for k in ("X", "y")})
bitacora("P-06", "Preprocesar: RGB → 224×224 → JPG q90 → [0,1]; augmentation definido solo para train", f"{len(df)} imágenes limpias", "Arrays 224×224×3 en [0,1]; augmentation solo en train; archivos JPG q90 guardados",
         f"train {D['train']['X'].shape}, val {D['val']['X'].shape}, test {D['test']['X'].shape}; rango [0,1] verificado", "Ninguna: los metadatos de archivo que se pierden al redimensionar están fuera del alcance de la conformidad", "preprocesado/, parametros_preprocesamiento.json, preprocesamiento_antes_despues.png")
```

```python
# ---------- Utilidades de métricas (se usan en los Pasos 7, 9 y 10) ----------
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, balanced_accuracy_score, roc_auc_score, confusion_matrix
def metricas(y, prob, thr=0.5):
    """Clase positiva = Conforme (1). FP = foto NO conforme aceptada como conforme (falso positivo normativo)."""
    y = np.asarray(y).astype(int); prob = np.asarray(prob, dtype=float); pred = (prob >= thr).astype(int)
    tn, fp, fn, tp = confusion_matrix(y, pred, labels=[0, 1]).ravel()
    return dict(accuracy=accuracy_score(y, pred), precision=precision_score(y, pred, zero_division=0), recall=recall_score(y, pred, zero_division=0),
                f1=f1_score(y, pred, zero_division=0), balanced_accuracy=balanced_accuracy_score(y, pred), specificity=(tn / (tn + fp)) if (tn + fp) else np.nan,
                auc_roc=roc_auc_score(y, prob) if len(set(y)) > 1 else np.nan, tp=int(tp), tn=int(tn), fp=int(fp), fn=int(fn))
```

## PASO 7 — Entrenar ResNet-50 y VGG-16 (Protocolo P05–P07, P09 · T07–T09, T11)
**Qué hace:** para cada arquitectura y cada semilla entrena una CNN **real** con pesos **ImageNet** (transfer learning):
* **Etapa 1:** base congelada, se entrena solo la cabeza (GlobalAveragePooling → Dropout → Dense sigmoide), LR 1e-4.
* **Etapa 2 (fine-tuning / descongelamiento progresivo):** se descongelan las capas superiores (BatchNorm queda congelada), LR 1e-5.
* Fijo del protocolo: Adam, decaimiento exponencial, batch 32, **máx. 50 épocas**, **early stopping paciencia 10 sobre `val_loss`**, pérdida binaria con **pesos de clase**, selección por **menor `val_loss`** (el *test* no participa).
* Cada corrida recibe un ID `EXP_<ARQ>_S<semilla>_2026` y guarda: configuración, historial, curvas, checkpoint, predicciones (val y test, con probabilidad) y tiempo de inferencia por imagen.

**Decisiones que el protocolo NO definía (las tomo aquí y quedan registradas como incidencia):** LR de la etapa de fine-tuning (1e-5), cuántas capas descongelar (30 en ResNet-50, 4 en VGG-16) y el reparto de épocas entre etapas (25 + 25).

**Semillas:** el protocolo pide 5 (10, 20, 30, 40, 50). Para que el piloto sea rápido, aquí se usan **2** (10 y 20): cambia `SEMILLAS` si quieres las 5. ⏱ Con GPU T4 espera ~10–20 min. *La GPU no es 100 % determinista: repetir una corrida puede variar levemente; por eso las predicciones se guardan en archivo.*


```python
# ---------- PASO 7: entrenamiento (requiere TensorFlow + GPU recomendada) ----------
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import ResNet50, VGG16
print("TensorFlow", tf.__version__, "| GPU:", tf.config.list_physical_devices("GPU") or "NO HAY GPU (será lento)")

SEMILLAS = [10, 20]                         # protocolo completo: [10, 20, 30, 40, 50]
HP = dict(batch=32, lr_etapa1=1e-4, lr_etapa2=1e-5, decay_rate=0.9, decay_cada_epocas=5, max_epocas=50, ep_etapa1=25, paciencia=10, dropout=0.3)
NFT = dict(ResNet50=30, VGG16=4)            # capas superiores descongeladas en la etapa 2
MEAN_BGR = np.array([103.939, 116.779, 123.68], dtype="float32")
def pre_caffe(x01):                         # equivale a preprocess_input de ResNet50/VGG16 (modo caffe); parte de [0,1] según protocolo
    return ((x01 * 255.0)[..., ::-1] - MEAN_BGR).astype("float32")

class SeqAug(keras.utils.Sequence):
    """Lotes de entrenamiento con augmentation (solo train) y pesos de clase como sample_weight."""
    def __init__(self, X, y, batch, seed, pesos):
        super().__init__(); self.X, self.y, self.b, self.rng, self.pesos = X, y, batch, np.random.RandomState(seed), pesos
        self.orden = np.arange(len(X)); self.rng.shuffle(self.orden)
    def __len__(self): return int(np.ceil(len(self.X) / self.b))
    def __getitem__(self, i):
        idx = self.orden[i * self.b:(i + 1) * self.b]
        xb = np.stack([augmentar(self.X[j], self.rng) for j in idx])
        return pre_caffe(xb), self.y[idx], np.array([self.pesos[int(self.y[j])] for j in idx], dtype="float32")
    def on_epoch_end(self): self.rng.shuffle(self.orden)

class MejorValLoss(keras.callbacks.Callback):
    def __init__(self): super().__init__(); self.best, self.w, self.ep = np.inf, None, -1
    def on_epoch_end(self, epoch, logs=None):
        if logs["val_loss"] < self.best: self.best, self.w, self.ep = logs["val_loss"], self.model.get_weights(), epoch

def construir(nombre):
    base = (ResNet50 if nombre == "ResNet50" else VGG16)(weights="imagenet", include_top=False, input_shape=(224, 224, 3)); base.trainable = False
    inp = keras.Input((224, 224, 3)); x = base(inp, training=False)
    x = layers.GlobalAveragePooling2D()(x); x = layers.Dropout(HP["dropout"])(x); out = layers.Dense(1, activation="sigmoid")(x)
    return keras.Model(inp, out), base
def compilar(model, lr0, pasos_por_epoca):
    sched = keras.optimizers.schedules.ExponentialDecay(lr0, decay_steps=pasos_por_epoca * HP["decay_cada_epocas"], decay_rate=HP["decay_rate"])
    model.compile(optimizer=keras.optimizers.Adam(sched), loss="binary_crossentropy", metrics=["accuracy"])

def entrenar(nombre, seed):
    exp = f"EXP_{nombre.upper()}_S{seed:03d}_2026"; inicio = _ahora(); t0 = time.time()
    try: tf.config.experimental.reset_memory_stats("GPU:0")
    except Exception: pass
    keras.utils.set_random_seed(seed)
    Xtr, ytr, Xva, yva, Xte, yte = D["train"]["X"], D["train"]["y"], D["val"]["X"], D["val"]["y"], D["test"]["X"], D["test"]["y"]
    n0, n1 = int((ytr == 0).sum()), int((ytr == 1).sum()); pesos = {0: len(ytr) / (2 * n0), 1: len(ytr) / (2 * n1)}
    model, base = construir(nombre); gen = SeqAug(Xtr, ytr, HP["batch"], seed, pesos); val = (pre_caffe(Xva), yva)
    hist, mejor = [], dict(val_loss=np.inf, w=None, etapa=0)
    for etapa, (lr0, max_ep) in enumerate([(HP["lr_etapa1"], HP["ep_etapa1"]), (HP["lr_etapa2"], HP["max_epocas"] - HP["ep_etapa1"])], start=1):
        if etapa == 2:
            base.trainable = True
            for l in base.layers[:-NFT[nombre]]: l.trainable = False
            for l in base.layers:
                if isinstance(l, layers.BatchNormalization): l.trainable = False
        compilar(model, lr0, len(gen)); mv = MejorValLoss()
        es = keras.callbacks.EarlyStopping(monitor="val_loss", patience=HP["paciencia"], restore_best_weights=False)
        h = model.fit(gen, validation_data=val, epochs=max_ep, callbacks=[mv, es], verbose=0)
        hh = pd.DataFrame(h.history); hh["etapa"] = etapa; hh["epoca"] = range(1, len(hh) + 1); hist.append(hh)
        print(f"  {exp} etapa {etapa}: {len(hh)} épocas, mejor val_loss={mv.best:.4f} (época {mv.ep + 1})")
        if mv.best < mejor["val_loss"]: mejor = dict(val_loss=mv.best, w=mv.w, etapa=etapa)
        model.set_weights(mv.w)                                              # cada etapa parte de su mejor punto
    model.set_weights(mejor["w"]); t_train = time.time() - t0               # criterio de selección: menor val_loss
    hist = pd.concat(hist, ignore_index=True); hist.to_csv(f"{EVI}/historial_{exp}.csv", index=False)
    fig, ax = plt.subplots(1, 2, figsize=(10, 3.2)); hist.reset_index(drop=True)[["loss", "val_loss"]].plot(ax=ax[0], title=f"{exp} · pérdida"); hist.reset_index(drop=True)[["accuracy", "val_accuracy"]].plot(ax=ax[1], title="exactitud")
    plt.tight_layout(); plt.savefig(f"{FIG}/curvas_{exp}.png", dpi=110); plt.close()
    try:
        model.save_weights(f"{MOD}/{exp}.weights.h5"); model.save(f"{MOD}/{exp}.keras")
    except Exception as e: print("  (aviso) no se pudo guardar un checkpoint:", e)
    pva = model.predict(pre_caffe(Xva), verbose=0, batch_size=32).ravel()
    _ = model(tf.constant(pre_caffe(Xte[:1])), training=False)             # calentamiento
    pte, ms = [], []
    for i in range(len(Xte)):
        t1 = time.perf_counter(); p = model(tf.constant(pre_caffe(Xte[i:i + 1])), training=False).numpy().ravel()[0]; ms.append((time.perf_counter() - t1) * 1000); pte.append(float(p))
    pte = np.array(pte)
    filas = []
    for s, M, p, tms in [("val", D["val"]["meta"], pva, [np.nan] * len(pva)), ("test", D["test"]["meta"], pte, ms)]:
        for k, (_, r) in enumerate(M.iterrows()):
            filas.append(dict(exp_id=exp, image_id=r.image_id, split=s, true_class=int(r.clase), predicted_class=int(p[k] >= .5), prob_conforme=float(p[k]), model=nombre, seed=seed, inference_ms=tms[k], subtipo=r.subtipo, familia=r.familia))
    pd.DataFrame(filas).to_csv(f"{EVI}/predictions_{nombre}_s{seed}.csv", index=False)
    mv_, mt_ = metricas(yva, pva), metricas(yte, pte)
    cfg = dict(exp_id=exp, modelo=nombre, semilla=seed, pesos_iniciales="imagenet", hiperparametros=HP, capas_descongeladas_etapa2=NFT[nombre], optimizador="Adam + ExponentialDecay", perdida="binary_crossentropy con pesos de clase",
               pesos_de_clase={str(k): round(v, 4) for k, v in pesos.items()}, entrada=[224, 224, 3], parametros=int(model.count_params()), etapa_ganadora=mejor["etapa"], inicio=inicio, fin=_ahora(), tensorflow=tf.__version__)
    json.dump(cfg, open(f"{EVI}/config_{nombre}_s{seed}.json", "w"), indent=2, ensure_ascii=False)
    try: gpu_pico = tf.config.experimental.get_memory_info("GPU:0")["peak"] / 1e6
    except Exception: gpu_pico = np.nan
    try:
        import psutil; ram_mb = psutil.Process().memory_info().rss / 1e6
    except Exception: ram_mb = np.nan
    reg_log(f"{exp}: {len(hist)} épocas, {t_train:.0f} s, GPU pico {gpu_pico:.0f} MB, RAM {ram_mb:.0f} MB")
    return dict(exp_id=exp, gpu_pico_mb=gpu_pico, ram_mb=ram_mb, model=nombre, seed=seed, best_val_loss=mejor["val_loss"], etapa_ganadora=mejor["etapa"], epocas_totales=len(hist), tiempo_entrenamiento_s=round(t_train, 1), parametros=int(model.count_params()),
                inferencia_ms_media=float(np.mean(ms)), **{f"val_{k}": v for k, v in mv_.items()}, **{f"test_{k}": v for k, v in mt_.items()})

resultados, errores = [], []
for nombre in ["ResNet50", "VGG16"]:
    for seed in SEMILLAS:
        try:
            print(f"▶ {nombre} semilla {seed}"); resultados.append(entrenar(nombre, seed))
        except Exception as e:
            errores.append(f"{nombre} s{seed}: {type(e).__name__}: {e}"); print("  ✖ FALLÓ:", errores[-1])
assert resultados, "Ninguna corrida terminó: revisa los errores de arriba"
RPS = pd.DataFrame(resultados); RPS.to_csv(f"{EVI}/resultados_por_semilla.csv", index=False)
json.dump(dict(corridas_ok=len(resultados), corridas_fallidas=errores, semillas=SEMILLAS, hiperparametros=HP), open(f"{EVI}/resumen_entrenamiento.json", "w"), indent=2, ensure_ascii=False)
print(RPS[["exp_id", "best_val_loss", "epocas_totales", "tiempo_entrenamiento_s", "val_accuracy", "test_accuracy", "test_recall", "test_specificity", "test_auc_roc"]].round(3).to_string(index=False))
bitacora("P-07", "Entrenar ResNet-50 y VGG-16 (transfer learning + fine-tuning) × semillas", f"train={len(D['train']['X'])} / val={len(D['val']['X'])} / test={len(D['test']['X'])}", "Corridas completas sin errores; selección por menor val_loss; todas las métricas calculadas",
         f"{len(resultados)} corridas OK, {len(errores)} fallidas; épocas totales {RPS.epocas_totales.tolist()}", "Decisiones no definidas en el protocolo: LR de fine-tuning, capas descongeladas, reparto de épocas" if not errores else "; ".join(errores),
         "config_*.json, historial_*.csv, predictions_*.csv, resultados_por_semilla.csv, curvas_*.png, modelos/")
incidencia("INC-HP", "El protocolo no define LR de la etapa de fine-tuning, número de capas a descongelar ni reparto de épocas entre etapas", "El V1.3 fija un LR único (1e-4) pero también pide descongelamiento progresivo", "Metodológica", "No (se documentó)",
           "Fijar en V1.4: etapa 1 LR 1e-4 (cabeza), etapa 2 LR 1e-5 (30 capas ResNet-50 / 4 capas VGG-16), 25+25 épocas, BatchNorm congelada.", "config_*.json")
```

## PASO 9 — Métricas y análisis estadístico (Protocolo P10–P11 · T11–T12)
**Qué hace:**
1. **Métricas por corrida** (Accuracy, Precision, Recall, F1, Balanced Accuracy, AUC-ROC, Specificity, matriz de confusión, nº de parámetros, tiempo de inferencia) y **media ± desviación estándar por arquitectura**.
2. **Selección del modelo final por arquitectura** usando **solo `val_loss`** (T09). Los modelos seleccionados son los que se comparan.
3. **McNemar exacto** (mismo conjunto de imágenes, α = 0.05) con **odds ratio** como tamaño de efecto, para las **3 comparaciones** (manual vs ResNet-50, manual vs VGG-16, ResNet-50 vs VGG-16) y **corrección Holm-Bonferroni** por comparaciones múltiples.
4. **Tiempos:** Shapiro-Wilk sobre las diferencias pareadas → **t de Student pareada** si es normal, **Wilcoxon** si no.
5. **Chequeo de hipótesis** H1, H2 y H3 de la tesis (con intervalo de confianza de Clopper-Pearson para la tasa de falsos positivos normativos).

**Cómo leer el odds ratio:** en McNemar, OR = *b / c*, con *b* = fotos que acierta solo el método A y *c* = fotos que acierta solo el método B. Si algún valor es 0 se aplica la corrección de Haldane-Anscombe (+0.5). Con ~22 fotos de prueba **todo p-value será poco potente**: es normal; este paso valida que el código funciona de punta a punta.


```python
# ---------- PASO 9: métricas + estadística ----------
from scipy import stats
RPS = pd.read_csv(f"{EVI}/resultados_por_semilla.csv")
modelos = sorted(RPS.model.unique())
cols = ["accuracy", "precision", "recall", "f1", "balanced_accuracy", "specificity", "auc_roc"]
filas = []
for m in modelos:
    s = RPS[RPS.model == m]
    fila = dict(modelo=m, n_corridas=len(s), parametros=int(s.parametros.iloc[0]), inferencia_ms=f"{s.inferencia_ms_media.mean():.1f}")
    for c in cols: fila[c] = f"{s['test_'+c].mean():.3f} ± {s['test_'+c].std(ddof=0):.3f}"
    filas.append(fila)
print("TEST — media ± desviación estándar entre semillas"); tabla_arch = pd.DataFrame(filas); print(tabla_arch.to_string(index=False)); tabla_arch.to_csv(f"{EVI}/metricas_por_arquitectura.csv", index=False)

# --- selección por menor val_loss (el test no participa) ---
SEL = RPS.loc[RPS.groupby("model").best_val_loss.idxmin()].set_index("model")
print("\nModelo seleccionado por arquitectura (menor val_loss):"); print(SEL[["exp_id", "best_val_loss", "test_accuracy", "test_recall", "test_specificity"]].round(3).to_string())
P = {m: pd.read_csv(f"{EVI}/predictions_{m}_s{int(SEL.loc[m, 'seed'])}.csv").query("split == 'test'").set_index("image_id") for m in modelos}
man = pd.read_csv(f"{EVI}/manual_eval.csv" if os.path.exists(f"{EVI}/manual_eval.csv") else f"{EVI}/manual_eval_SIMULADO.csv").set_index("image_id")
MODO_MANUAL = man.modo.iloc[0]
ids = man.index.intersection(P[modelos[0]].index)
y = man.loc[ids, "clase"].astype(int).values
acierto = {"Manual": (man.loc[ids, "decision_manual"].values == y).astype(int)}
for m in modelos: acierto[m] = (P[m].loc[ids, "predicted_class"].values == y).astype(int)

def mcnemar_exacto(a, b):
    n01 = int(((a == 1) & (b == 0)).sum()); n10 = int(((a == 0) & (b == 1)).sum())      # b: solo A acierta | c: solo B acierta
    p = 1.0 if n01 + n10 == 0 else float(stats.binomtest(n01, n01 + n10, 0.5).pvalue)
    orr = (n01 + .5) / (n10 + .5) if (n01 == 0 or n10 == 0) else n01 / n10
    return dict(solo_A_acierta=n01, solo_B_acierta=n10, p_valor=p, odds_ratio=orr, corregido_haldane=bool(n01 == 0 or n10 == 0))
def holm(ps):
    o = np.argsort(ps); aj = np.empty(len(ps)); mx = 0
    for r, i in enumerate(o): mx = max(mx, (len(ps) - r) * ps[i]); aj[i] = min(1, mx)
    return aj
pares = [("Manual", modelos[0]), ("Manual", modelos[1]), (modelos[0], modelos[1])]
comp = []
for A, B in pares:
    r = mcnemar_exacto(acierto[A], acierto[B]); r.update(A=A, B=B, n=len(ids), acierto_A=acierto[A].mean(), acierto_B=acierto[B].mean()); comp.append(r)
adj = holm(np.array([c["p_valor"] for c in comp]))
for c, a in zip(comp, adj): c["p_holm"] = float(a); c["significativo_alfa_0_05"] = bool(a < 0.05)
COMP = pd.DataFrame(comp)[["A", "B", "n", "acierto_A", "acierto_B", "solo_A_acierta", "solo_B_acierta", "odds_ratio", "corregido_haldane", "p_valor", "p_holm", "significativo_alfa_0_05"]]
print("\nMcNemar exacto + odds ratio (3 comparaciones) con Holm-Bonferroni:"); print(COMP.round(3).to_string(index=False))

# --- tiempos (pareados por imagen): manual (s) vs automatizado (s) ---
t_man = man.loc[ids, "tiempo_seg"].values
TIEMPOS = {}
fig, ax = plt.subplots(figsize=(6, 3.6)); datos, etq = [t_man], ["Manual (s)"]
for m in modelos:
    t_aut = P[m].loc[ids, "inference_ms"].values / 1000.0; d = t_man - t_aut
    sw = stats.shapiro(d) if len(d) >= 3 else (np.nan, np.nan); normal = bool(sw[1] > 0.05)
    prueba = "t pareada" if normal else "Wilcoxon"
    pv = float(stats.ttest_rel(t_man, t_aut).pvalue) if normal else float(stats.wilcoxon(t_man, t_aut).pvalue)
    TIEMPOS[m] = dict(media_manual_s=float(t_man.mean()), media_auto_s=float(t_aut.mean()), reduccion_pct=float(100 * (1 - t_aut.mean() / t_man.mean())), shapiro_W=float(sw[0]), shapiro_p=float(sw[1]), normal=normal, prueba=prueba, p_valor=pv)
    datos.append(t_aut); etq.append(f"{m} (s)")
print("\nTiempos pareados:"); print(pd.DataFrame(TIEMPOS).T.round(4).to_string())
ax.boxplot(datos, labels=etq); ax.set_yscale("log"); ax.set_ylabel("segundos por foto (escala log)"); ax.set_title(f"Tiempo por foto — manual ({MODO_MANUAL}) vs CNN"); plt.tight_layout(); plt.savefig(f"{FIG}/tiempo_manual_vs_automatizado.png", dpi=120); plt.show()

# --- matrices de confusión + ROC de los modelos seleccionados ---
fig, ax = plt.subplots(1, len(modelos) + 1, figsize=(5 * (len(modelos) + 1), 4))
for a, (nom, pred, prob) in zip(ax, [(m, P[m].loc[ids, "predicted_class"].values, P[m].loc[ids, "prob_conforme"].values) for m in modelos]):
    cm = confusion_matrix(y, pred, labels=[0, 1]); a.imshow(cm, cmap="Blues"); a.set_title(f"{nom} (test)"); a.set_xticks([0, 1], ["No conf.", "Conf."]); a.set_yticks([0, 1], ["No conf.", "Conf."]); a.set_xlabel("predicho"); a.set_ylabel("real")
    for (i, j), v in np.ndenumerate(cm): a.text(j, i, v, ha="center", va="center", fontsize=14)
from sklearn.metrics import roc_curve
for m in modelos:
    fpr, tpr, _ = roc_curve(y, P[m].loc[ids, "prob_conforme"].values); ax[-1].plot(fpr, tpr, label=f"{m} (AUC={roc_auc_score(y, P[m].loc[ids, 'prob_conforme'].values):.2f})")
ax[-1].plot([0, 1], [0, 1], "k--"); ax[-1].set_title("Curvas ROC (test)"); ax[-1].set_xlabel("FPR"); ax[-1].set_ylabel("TPR"); ax[-1].legend()
plt.tight_layout(); plt.savefig(f"{FIG}/confusion_y_roc.png", dpi=120); plt.show()

# --- hipótesis (con n pequeño: solo valida la mecánica) ---
def cp_sup(k, n, alfa=0.05): return float(stats.beta.ppf(1 - alfa, k + 1, n - k)) if k < n else 1.0
H = {}
for m in modelos:
    mt = metricas(y, P[m].loc[ids, "prob_conforme"].values); n_neg = mt["tn"] + mt["fp"]
    pcomp = COMP[(COMP.A == "Manual") & (COMP.B == m)].iloc[0]
    H[m] = dict(H1_accuracy=mt["accuracy"], H1_recall=mt["recall"], H1_cumple_umbral_95=bool(mt["accuracy"] >= .95 and mt["recall"] >= .95), H1_mcnemar_vs_manual_p_holm=float(pcomp.p_holm),
                H2_reduccion_tiempo_pct=TIEMPOS[m]["reduccion_pct"], H2_cumple_50pct=bool(TIEMPOS[m]["reduccion_pct"] >= 50 and TIEMPOS[m]["p_valor"] < .05),
                H3_fp_normativos=mt["fp"], H3_tasa_fp=(mt["fp"] / n_neg) if n_neg else np.nan, H3_ic95_superior=cp_sup(mt["fp"], n_neg), H3_cumple_cero_fp=bool(mt["fp"] == 0))
print("\nChequeo de hipótesis (PILOTO, n =", len(ids), "):"); print(pd.DataFrame(H).T.round(3).to_string())
json.dump(dict(n_test=len(ids), modo_manual=MODO_MANUAL, modelos_seleccionados=SEL.exp_id.to_dict(), comparaciones_pareadas=COMP.to_dict("records"), tiempos=TIEMPOS, hipotesis=H, correccion="Holm-Bonferroni (3 comparaciones)", alfa=0.05),
          open(f"{EVI}/resultado_analisis_estadistico.json", "w"), indent=2, ensure_ascii=False, default=lambda o: o.item() if hasattr(o, "item") else str(o))
COMP.to_csv(f"{EVI}/tabla_resultados_estadisticos.csv", index=False)
bitacora("P-09", "Métricas, McNemar exacto + odds ratio (3 comparaciones), Holm-Bonferroni, Shapiro-Wilk, t/Wilcoxon", f"{len(ids)} fotos de test: manual ({MODO_MANUAL}) + {len(modelos)} CNN seleccionadas por val_loss",
         "Código estadístico completo sin errores; odds ratio y corrección múltiple calculados", f"p Holm = {[round(c['p_holm'], 3) for c in comp]}; OR = {[round(c['odds_ratio'], 2) for c in comp]}; tiempos: {({m: TIEMPOS[m]['prueba'] for m in modelos})}",
         "n de test pequeño → pruebas con muy poca potencia" , "resultado_analisis_estadistico.json, tabla_resultados_estadisticos.csv, metricas_por_arquitectura.csv")
```
